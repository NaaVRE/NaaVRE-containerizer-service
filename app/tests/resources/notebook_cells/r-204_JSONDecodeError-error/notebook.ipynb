```
# Create a JSON
library(jsonlite)

# Create a list (equivalent to a Python dictionary)
data <- list(
  task = "Write Python",
  number = 1,
  skills = c("python", "sql")
)

# Serialize with toJSON()
json_str <- toJSON(data, auto_unbox = TRUE)

# Print the JSON string
print(json_str)
```

```
# Use the JSON

# Create the original list
data <- list(
  task = "Write Python",
  number = 1,
  skills = c("python", "sql")
)

# Deserialize JSON using fromJSON()
deserialized_data <- fromJSON(json_str)

# Print the deserialized data
print(deserialized_data)

# Assert that the original and deserialized data are equal
stopifnot(isTRUE(all.equal(data, deserialized_data)))
```
